### **Isotope Geochemistry**

# Oxygen isotope notations

Oxygen has three stable isotopes:

- $^{16}\mathrm{O}$
- $^{17}\mathrm{O}$
- $^{18}\mathrm{O}$

In stable isotope geochemistry, we usually compare the ratio of a heavy isotope to the most abundant isotope, $^{16}\mathrm{O}$.

For oxygen isotopes:

$$
R^{17} = \frac{^{17}\mathrm{O}}{^{16}\mathrm{O}}
$$

$$
R^{18} = \frac{^{18}\mathrm{O}}{^{16}\mathrm{O}}
$$

Because absolute isotope ratios are very small and difficult to compare directly, isotope compositions are commonly reported relative to a reference material, such as VSMOW.

The delta notation is defined as:

$$
\delta^{x}\mathrm{O}
=
\left(
\frac{R^{x}_{\mathrm{sample}}}{R^{x}_{\mathrm{reference}}}
-1
\right)
\times 1000
$$

where $x$ can be 17 or 18.

For example:

$$
\delta^{18}\mathrm{O}
=
\left(
\frac{
\left(^{18}\mathrm{O}/^{16}\mathrm{O}\right)_{\mathrm{sample}}
}{
\left(^{18}\mathrm{O}/^{16}\mathrm{O}\right)_{\mathrm{reference}}
}
-1
\right)
\times 1000
$$

The result is reported in per mil, written as ‰.

A positive $\delta$ value means that the sample has a higher heavy-isotope/light-isotope ratio than the reference.

A negative $\delta$ value means that the sample has a lower heavy-isotope/light-isotope ratio than the reference.

# Linearized delta notation: $\delta'$ notation

In triple oxygen isotope geochemistry, we often use the linearized form of delta notation.

This is called delta-prime notation:

$$
\delta'^{x}\mathrm{O}
=
1000
\ln
\left(
1+
\frac{\delta^{x}\mathrm{O}}{1000}
\right)
$$

where $x$ can be 17 or 18.

Therefore:

$$
\delta'^{17}\mathrm{O}
=
1000
\ln
\left(
1+
\frac{\delta^{17}\mathrm{O}}{1000}
\right)
$$

and

$$
\delta'^{18}\mathrm{O}
=
1000
\ln
\left(
1+
\frac{\delta^{18}\mathrm{O}}{1000}
\right)
$$

For small isotope variations, $\delta$ and $\delta'$ are very similar.

However, $\delta'$ is mathematically useful because isotope fractionation becomes additive in logarithmic space.

# Fractionation factor

The isotope fractionation factor between two phases, A and B, is defined as:

$$
\alpha^{x}_{A-B}
=
\frac{R^{x}_{A}}{R^{x}_{B}}
$$

where $R^{x}$ is the isotope ratio of isotope $x$ relative to $^{16}\mathrm{O}$.

For oxygen isotopes:

$$
\alpha^{18}_{A-B}
=
\frac{
\left(^{18}\mathrm{O}/^{16}\mathrm{O}\right)_A
}{
\left(^{18}\mathrm{O}/^{16}\mathrm{O}\right)_B
}
$$

The fractionation factor can also be calculated from delta values:

$$
\alpha^{x}_{A-B}
=
\frac{
1+\delta^{x}O_A/1000
}{
1+\delta^{x}O_B/1000
}
$$

In isotope geochemistry, fractionation is often reported as:

$$
1000 \ln \alpha^{x}_{A-B}
$$

Using delta-prime notation:

$$
1000 \ln \alpha^{x}_{A-B}
=
\delta'^{x}O_A
-
\delta'^{x}O_B
$$

This is one reason why the logarithmic notation is useful.

# Triple oxygen isotope relationship

In mass-dependent fractionation, $^{17}\mathrm{O}$ and $^{18}\mathrm{O}$ usually vary together.

Because $^{17}\mathrm{O}$ is intermediate in mass between $^{16}\mathrm{O}$ and $^{18}\mathrm{O}$, the change in $\delta^{17}\mathrm{O}$ is approximately about half of the change in $\delta^{18}\mathrm{O}$.

In linearized notation, this relationship is commonly written as:

$$
\delta'^{17}\mathrm{O}
=
\lambda
\delta'^{18}\mathrm{O}
+
\gamma
$$

where:

- $\lambda$ is the slope of the reference line
- $\gamma$ is the intercept
- $\delta'^{17}\mathrm{O}$ and $\delta'^{18}\mathrm{O}$ are the linearized isotope values

A commonly used simplified form is:

$$
\delta'^{17}\mathrm{O}
=
\lambda
\delta'^{18}\mathrm{O}
$$

with, for example:

$$
\lambda = 0.528
$$

This line represents the expected mass-dependent relationship between $^{17}\mathrm{O}$ and $^{18}\mathrm{O}$.

# Capital delta (Cap Delta) notation: $\Delta'^{17}\mathrm{O}$

The parameter $\Delta'^{17}\mathrm{O}$ describes the deviation of a sample from a chosen mass-dependent reference line.

It is commonly defined as:

$$
\Delta'^{17}\mathrm{O}
=
\delta'^{17}\mathrm{O}
-
\lambda
\delta'^{18}\mathrm{O}
$$

where $\lambda$ is the chosen reference slope.

If the reference line includes an intercept, the equation becomes:

$$
\Delta'^{17}\mathrm{O}
=
\delta'^{17}\mathrm{O}
-
\left(
\lambda
\delta'^{18}\mathrm{O}
+
\gamma
\right)
$$

If $\Delta'^{17}\mathrm{O} = 0$, the sample lies exactly on the chosen reference line.

If $\Delta'^{17}\mathrm{O} > 0$, the sample lies above the reference line.

If $\Delta'^{17}\mathrm{O} < 0$, the sample lies below the reference line.

Because $\Delta'^{17}\mathrm{O}$ values are usually very small, they are often reported in per meg:

$$
\Delta'^{17}\mathrm{O}_{\mathrm{per\ meg}}
=
1000
\left[
\delta'^{17}\mathrm{O}
-
\lambda
\delta'^{18}\mathrm{O}
\right]
$$

Here, $\delta'^{17}\mathrm{O}$ and $\delta'^{18}\mathrm{O}$ are in per mil, while $\Delta'^{17}\mathrm{O}_{\mathrm{per\ meg}}$ is in per meg.

# $\theta$ and $\lambda$

In triple oxygen isotope geochemistry, both $\theta$ and $\lambda$ describe slopes in oxygen isotope space, but they are often used differently.

The exponent $\theta$ is commonly used for a specific fractionation process between two phases:

$$
\theta_{A-B}
=
\frac{
\ln \alpha^{17}_{A-B}
}{
\ln \alpha^{18}_{A-B}
}
$$

For example, $\theta$ may describe equilibrium fractionation between quartz and water, or another specific mineral-fluid pair.

The parameter $\lambda$ is commonly used for a reference line or empirical relationship in isotope space:

$$
\delta'^{17}\mathrm{O}
=
\lambda
\delta'^{18}\mathrm{O}
+
\gamma
$$

In simple terms:

- $\theta$ describes the slope of a specific fractionation process.
- $\lambda$ describes the slope of a chosen reference line.

Because $\Delta'^{17}\mathrm{O}$ depends on the chosen reference slope, the value of $\lambda$ should always be reported.

### **Hyperbolic Mixing Equation**
The isotopic ratio of the mixture is given by:

$$
\left( \frac{^{87}\text{Sr}}{^{86}\text{Sr}} \right)_{\text{mix}} = \frac{[\text{Sr}]_A \cdot \left( \frac{^{87}\text{Sr}}{^{86}\text{Sr}} \right)_A \cdot (1 - x) + [\text{Sr}]_B \cdot \left( \frac{^{87}\text{Sr}}{^{86}\text{Sr}} \right)_B \cdot x \cdot r}{[\text{Sr}]_A \cdot (1 - x) + [\text{Sr}]_B \cdot x \cdot r}
$$

Where:
- $[\text{Sr}]_A$, $[\text{Sr}]_B$ = Sr concentrations in end-members A (mantle) and B (crust)
- $r = \frac{[\text{Sr}]_B}{[\text{Sr}]_A}$ = concentration ratio (5:1, 1:10, etc.)
- $x$ = mixing fraction (0 to 1)

$$ \delta^{18}\text{O}_{\text{mix}} = \delta^{18}\text{O}_A (1 - x) + \delta^{18}\text{O}_B \cdot x \ \ \text{(linear mixing for oxygen)} $$

```python
# Define end-members (A = mantle, B = crust)
A = np.array([0.703, 6.0])   # Low 87Sr/86Sr, low δ18O (mantle)
B = np.array([0.710, 12.0])  # High 87Sr/86Sr, high δ18O (crust)

# Mixing parameter (fraction of contaminant, x)
x = np.linspace(0, 1, 100)

# Sr concentration ratios (Sr_B/Sr_A) to test
ratios = [0.2, 0.5, 1, 2, 10]  # 5:1, 2:1, 1:1, 1:2, 1:10

plt.figure(figsize=(8,6))

Sr_array = []
O_array = []

for i, ratio in enumerate(ratios):
    # Hyperbolic mixing for Sr isotopes
    Sr_mix = (A[0] * (1 - x) + B[0] * x * ratio) / (1 - x + x * ratio)
    Sr_array.append(Sr_mix)
    # Linear mixing for O isotopes (no fractionation)
    O_mix = A[1] * (1 - x) + B[1] * x
    O_array.append(O_mix)
    
    plt.plot(Sr_mix, O_mix, color="black", lw=2)


plt.fill_betweenx(O_array[0], Sr_array[1], Sr_array[0], color="black", alpha=0.15)
plt.fill_betweenx(O_array[4], Sr_array[4], Sr_array[3], color="black", alpha=0.15)
plt.text(A[0]-0.00025, A[1]-0.15, "A", fontsize=12, weight="bold")
plt.text(B[0]+0.00015, B[1]+0.02, "B", fontsize=12, weight="bold")
plt.text(0.7073, 7.1, "Source Contamination", fontsize=12, ha="center", rotation=20)
plt.text(0.7053, 8.72, "Crustal Contamination", fontsize=12, ha="center", rotation=35)
plt.text(0.7049, 10, "5:1", fontsize=12, ha="center", rotation=35)
plt.text(0.7055, 9.15, "2:1", fontsize=12, ha="center", rotation=35)
plt.text(0.7065, 9, "1:1", fontsize=12, ha="center", rotation=35)
plt.text(0.7070, 8.44, "1:2", fontsize=12, ha="center", rotation=35)
plt.text(0.7080, 7.2, "1:5", fontsize=12, ha="center", rotation=32)
plt.plot(A[0], A[1], "ko", markersize=8, label="Mantle (A)")
plt.plot(B[0], B[1], "ko", markersize=8, label="Crust (B)")
plt.plot(radiogenic_data['D2004-3']['87Sr/86Sr'], bulk_data['d18O'][2], "s", color='black', label="D2004, 3 & 11")
plt.plot(radiogenic_data['D2004-11']['87Sr/86Sr'], bulk_data['d18O'][1], "s", color='black')
plt.xlabel(r'$^{87}Sr/^{86}Sr$', fontsize=12)
plt.ylabel(r'$\delta^{18}O \ (\perthousand)$', fontsize=12)
plt.title("Hypothetical mixing diagram (James, 1981)", fontsize=14)
plt.legend()
plt.show()
```

## εNd hesabı

```python
def epsilon_nd(nd_143_144_sample):
    # CHUR reference value for 143Nd/144Nd
    nd_143_144_chur = 0.512638

    # εNd
    epsilon_nd = ((nd_143_144_sample / nd_143_144_chur) - 1) * 10**4

    return epsilon_nd
```

```python
plt.plot(epsilon_nd(radiogenic_data['D2004-3']['143Nd/144Nd']), bulk_data['d18O'][2], "o", color='black')
plt.plot(epsilon_nd(radiogenic_data['D2004-11']['143Nd/144Nd']), bulk_data['d18O'][1], "o", color='black')
plt.xlabel(r'eNd'), plt.ylabel(r'$\delta^{18}O$')
plt.title('d18O vs eNd')
plt.show()
```
