# Literature

```python
from pathlib import Path
from IPython.display import display, HTML

pdf_dir = Path("literature")
thumb_dir = Path("_static/literature_thumbs")

pdf_files = sorted(pdf_dir.glob("*.pdf"))

cards = []
for pdf_path in pdf_files:
    thumb_path = thumb_dir / f"{pdf_path.stem}.png"
    title = pdf_path.stem.replace("_", " ")

    card = f"""
    <div style="width:220px; text-align:center; margin:15px;">
        <a href="{pdf_path.as_posix()}" target="_blank" style="text-decoration:none; color:inherit;">
            <img src="{thumb_path.as_posix()}" style="width:180px; border:1px solid #ccc; box-shadow:2px 2px 8px rgba(0,0,0,0.15);">
            <div style="margin-top:10px; font-size:14px;">{title}</div>
        </a>
    </div>
    """
    cards.append(card)

html = f"""
<div style="display:flex; flex-wrap:wrap; gap:10px;">
    {''.join(cards)}
</div>
"""

display(HTML(html))
```
